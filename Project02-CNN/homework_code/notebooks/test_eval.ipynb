```python
"""
Notebook cell / standalone script for evaluating a saved model checkpoint.

Loads a trained .pth file, runs it on the full test set, and prints the
test accuracy. Useful for quick evaluation without re-training.
"""
import sys
import os

sys.path.append(os.path.abspath('../'))

import torch
from models.model import BaseModel
from data.data_loader import create_dataloaders

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
#chenge these two code line below
model_path = '../models/saved_models/best_model_A_20260430_WithAdam.pth'
#notice: dropout_rate should set with best config for each model 0.3  for A and 0.4 for B and C blocks
# check the config.yaml 
model = BaseModel(block_type='A', num_classes=10, dropout_rate=0.3) #set block_type
model.load_state_dict(torch.load(model_path, map_location=device))
model.to(device)
model.eval()

_, _, test_loader = create_dataloaders('../data/dataset', batch_size=256, num_workers=4)

correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = outputs.max(1)
        correct += predicted.eq(labels).sum().item()
        total += labels.size(0)

print(f"Block - Test Accuracy: {100.0 * correct / total:.2f}%")
```
